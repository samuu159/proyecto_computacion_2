Prediccion jugadores del equipo

```python
import pandas as pd
import re
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Cargar el CSV de jugadores desde la ruta 'Jugadores_Fantasy/Jugadores_Ordenados_Puntos.csv'
df_jugadores = pd.read_csv('Jugadores_Fantasy/Jugadores_Ordenados_Puntos.csv', thousands='.', decimal=',')

# Convertir la cadena 'info_jornadas' a una lista de Python
df_jugadores['info_jornadas'] = df_jugadores['info_jornadas'].apply(ast.literal_eval)

# Función para extraer los puntos de la siguiente jornada
def puntos_siguiente_jornada(info_jornadas):
    for jornada_info in info_jornadas:
        match = re.search(r'J\d+: Puntos (\d+)', jornada_info)
        if match:
            return int(match.group(1))
    return 0

# Crear una nueva columna 'puntos_siguiente_jornada' con valores numéricos
df_jugadores['puntos_siguiente_jornada'] = df_jugadores['info_jornadas'].apply(puntos_siguiente_jornada)

# Seleccionar características (X) y objetivos (y) para el entrenamiento del modelo
X_train = df_jugadores[['posicion', 'valor', 'media', 'partidos', 'goles', 'tarjetas']]
y_valor = df_jugadores['valor']
y_puntos_siguiente_jornada = df_jugadores['puntos_siguiente_jornada']

# Columnas categóricas para codificar
categorical_cols = ['posicion']

# Preprocesamiento para manejar variables categóricas
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), categorical_cols)
    ],
    remainder='passthrough'
)

# Aplicar preprocesamiento a las características
X_train_preprocessed = preprocessor.fit_transform(X_train)

# Inicializar y entrenar un modelo Random Forest para el valor y los puntos en la siguiente jornada
modelo_rf = RandomForestRegressor(n_estimators=100, random_state=42)
modelo_rf.fit(X_train_preprocessed, y_valor)  # Entrenar para el valor
modelo_rf_puntos_siguiente_jornada = RandomForestRegressor(n_estimators=100, random_state=42)
modelo_rf_puntos_siguiente_jornada.fit(X_train_preprocessed, y_puntos_siguiente_jornada)  # Entrenar para los puntos en la siguiente jornada

# Cargar el CSV del mercado desde la ruta 'mercado/mercado.csv'
df_mercado = pd.read_csv('team_data/Equipo.csv', thousands='.', decimal=',')

# Seleccionar características (X) para la predicción
X_prediccion = df_mercado[['posicion', 'valor', 'media', 'partidos', 'goles', 'tarjetas']]

# Aplicar preprocesamiento a las características de la predicción
X_prediccion_preprocessed = preprocessor.transform(X_prediccion)

# Hacer predicciones para el valor y los puntos en la siguiente jornada en el mercado
predicciones_valor_rf = modelo_rf.predict(X_prediccion_preprocessed)
predicciones_puntos_siguiente_jornada_rf = modelo_rf_puntos_siguiente_jornada.predict(X_prediccion_preprocessed)

# Agregar las predicciones de valor y puntos en la siguiente jornada al DataFrame del mercado
df_mercado['pred_valor'] = predicciones_valor_rf
df_mercado['pred_puntos_siguiente_jornada'] = predicciones_puntos_siguiente_jornada_rf

# Guardar el DataFrame modificado en un nuevo CSV
df_mercado.to_csv('team_data/Equipo_Predicciones.csv', index=False)
```
