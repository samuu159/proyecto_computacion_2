```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
# Configuración del navegador
options = Options()
options.add_argument('--disable-notifications')

# Ruta al controlador de Chrome (reemplaza con la ubicación de tu controlador)
chrome_driver_path = '../../../chromedriver-win64/chromedriver.exe'

# Crear una instancia de ChromeDriver
driver = webdriver.Chrome(executable_path=chrome_driver_path, options=options)

# Navegar a la URL
driver.get('https://mister.mundodeportivo.com')

# Aceptar cookies
try:
    WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.ID, 'didomi-notice-learn-more-button')))
    boton = driver.find_element_by_id('didomi-notice-learn-more-button')
    boton.click()

    botones = driver.find_elements_by_css_selector('.didomi-components-button.didomi-button.didomi-button-standard.standard-button')

    if botones:
        botones[0].location_once_scrolled_into_view
        botones[0].click()
except Exception as e:
    print(f"Error al aceptar las cookies: {str(e)}")

# Inicio de sesión o registro, si es necesario
try:
    boton = driver.find_element_by_xpath('//*[@id="app"]/div/div[2]/div[2]/button')
    boton.click()

    boton = driver.find_element_by_xpath('//*[@id="app"]/div/div[2]/div[2]/button')
    boton.click()

    boton = driver.find_element_by_xpath('//*[@id="app"]/div/div[2]/div[2]/button')
    boton.click()

    boton = driver.find_element_by_xpath('//*[@id="app"]/div/div[2]/div[2]/button')
    boton.click()

    boton = driver.find_element_by_xpath('//*[@id="app"]/div/div[2]/div/button[3]')
    boton.click()


    campo_input = driver.find_element_by_xpath('//*[@id="email"]')

    # Escribir texto en el campo de entrada
    campo_input.send_keys("ivangarciamoriones23@gmail.com")

    campo_input = driver.find_element_by_xpath('//*[@id="app"]/div/div[2]/div/form/div[2]/input')

    # Escribir texto en el campo de entrada
    campo_input.send_keys("Cloe2016!")

    boton = driver.find_element_by_xpath('//*[@id="app"]/div/div[2]/div/form/div[3]/button')
    boton.click()
except Exception as e:
    print(f"Error en el inicio de sesión o registro: {str(e)}")

# Interactuar con la lista de jugadores
try:
    WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.XPATH, '//*[@id="content"]/header/div[2]/ul/li[5]')))
    
    boton = driver.find_element_by_xpath('//*[@id="content"]/header/div[2]/ul/li[5]')
    boton.click()
    
    WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.XPATH, '//*[@id="content"]/div[2]/div[1]/button[2]')))
    
    boton = driver.find_element_by_xpath('//*[@id="content"]/div[2]/div[1]/button[2]')
    boton.click()
    
    WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.XPATH, '/html/body/div[6]/div[3]/div[3]/ul')))
    lista = driver.find_element(By.XPATH, '/html/body/div[6]/div[3]/div[3]/ul')
    jugadores = lista.find_elements(By.TAG_NAME, 'li')

    for jugador in jugadores:
        lista = driver.find_element(By.XPATH, '/html/body/div[6]/div[3]/div[3]/ul')
        jugadores = lista.find_elements(By.TAG_NAME, 'li')
        jugador.click()
        print('Clic en un jugador')
        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.XPATH, '/html/body/div[6]/div[3]/div[1]/div/div[1]/button')))
        boton = driver.find_element_by_xpath('/html/body/div[6]/div[3]/div[1]/div/div[1]/button')
        boton.click()
        time.sleep(2)
except Exception as e:
    print(f"Error al interactuar con la lista de jugadores: {str(e)}")

# Cierra el navegador
driver.quit()
```
