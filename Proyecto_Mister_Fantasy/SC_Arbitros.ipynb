# IMPORTS

```python
import csv
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.chrome.options import Options
from selenium.common.exceptions import NoSuchElementException
import time
```

# Fuente de los datos

https://es.besoccer.com/competicion/arbitros/primera

# Scraper

```python
# ********************************************** Para quitar notificaciones **********************************************
option = Options()
option.add_argument('--disable-notifications')
# ************************************************************************************************************************


# ********************************************** Ruta al controlador de Chrome *******************************************
driver = webdriver.Chrome("../chromedriver-mac-arm64/chromedriver")
url = "https://es.besoccer.com/competicion/arbitros/primera"
driver.get(url)
driver.maximize_window() # Maximizar la ventana del navegador
# ************************************************************************************************************************



# ********************************************** Aceptar botón cokies **********************************************
wait = WebDriverWait(driver, 10)
accept_button = wait.until(EC.element_to_be_clickable((By.XPATH, '//*[@id="qc-cmp2-ui"]/div[2]/div/button[3]')))
accept_button.click()
# ********************************************* ¡Fin botón aceptar cokies! ******************************************


# ********************************************** Sacar datos de los Árbitros **********************************************

csv_file_path = 'datos_arbitros.csv' # Nombre del CSV donde se almacenaran los datos

# Abrir el archivo CSV en modo de escritura. Si no existe, se creará.
with open(csv_file_path, mode='w', newline='', encoding='utf-8') as file:
    
    writer = csv.writer(file)
    # Escribir la fila de encabezado con los nombres de las columnas
    writer.writerow(['Árbitro', 'Partidos Jugados', 'Penaltis Pitados', 'Promedio Penalti', 'Tarjetas Amarillas', 'Promedio Amarilla', 'Tarjetas Rojas', 'Promedio Roja', 'Doble Amarilla', 'Promedio Doble Amarilla'])

    # ********************************************** COMIENZO DEL SACADO DE DATOS ********************************************
    for index in range(3, 23):
        xpath = f'//*[@id="mod_referees"]/div/table/tbody/tr[{index}]/td[2]/a'
        element = wait.until(EC.presence_of_element_located((By.XPATH, xpath)))
        driver.execute_script("arguments[0].scrollIntoView(true);", element)
        driver.execute_script("arguments[0].click();", element)

        arbitro            = wait.until(EC.presence_of_element_located((By.XPATH, '//*[@id="mod_detail_head_referee"]/div[1]/div/div[1]/div/h2'))).text
        partidos_jugados   = driver.find_element_by_xpath('//*[@id="mod_referee_stats"]/div/div[2]/div[2]/div/div/div[1]/div[2]').text
        penaltis_pitados   = driver.find_element_by_xpath('//*[@id="mod_referee_stats"]/div/div[2]/div[2]/div/div/div[2]/div[2]').text
        p_penalti          = driver.find_element_by_xpath('//*[@id="mod_referee_stats"]/div/div[2]/div[2]/div/div/div[2]/div[3]').text
        tarjetas_amarillas = driver.find_element_by_xpath('//*[@id="mod_referee_stats"]/div/div[2]/div[2]/div/div/div[3]/div[2]').text
        p_amarilla         = driver.find_element_by_xpath('//*[@id="mod_referee_stats"]/div/div[2]/div[2]/div/div/div[3]/div[3]').text
        tarjetas_rojas     = driver.find_element_by_xpath('//*[@id="mod_referee_stats"]/div/div[2]/div[2]/div/div/div[4]/div[2]').text
        p_roja             = driver.find_element_by_xpath('//*[@id="mod_referee_stats"]/div/div[2]/div[2]/div/div/div[4]/div[3]').text
        doble_amarilla     = driver.find_element_by_xpath('//*[@id="mod_referee_stats"]/div/div[2]/div[2]/div/div/div[5]/div[2]').text
        p_doble_amarilla   = driver.find_element_by_xpath('//*[@id="mod_referee_stats"]/div/div[2]/div[2]/div/div/div[5]/div[3]').text

  
        writer.writerow([arbitro, partidos_jugados, penaltis_pitados, p_penalti, tarjetas_amarillas, p_amarilla, tarjetas_rojas, p_roja, doble_amarilla, p_doble_amarilla])
        driver.back()
# ********************************************** FIN SACADO DEL DACADO DE DATOS **************************************************
```
